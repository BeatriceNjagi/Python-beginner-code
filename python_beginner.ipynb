```python
#28. Write a program that counts how many times each word appears in a list of words: ["apple","banana","apple","cherry","banana","banana"].
fruits = ["apple","banana","apple","cherry","banana","banana"]

fruit_count = {}   # empty dictionary

for fruit in fruits:
    if fruit in fruit_count:
        fruit_count[fruit] += 1
    else:
        fruit_count[fruit] = 1

print(fruit_count)
```

```python
#39. Given a list of numbers, ask the user to input the numbers, then compute: number ofpositive, negative, and zero values using conditionals.
count = int(input("How many numbers will you enter? "))

for i in range(count):
    num = int(input("Enter a number: "))
    numbers.append(num)

pos = neg = zero = 0

for n in numbers:
    if n > 0:
        pos += 1
    elif n < 0:
        neg += 1
    else:
        zero += 1

print("Positive:", pos)
print("Negative:", neg)
print("Zero:", zero)
```

```python
#40. Write a small program that asks the user to input their name, age, and a list of 3 favorite fruits (comma-separated). Store their data in a dictionary like {"name": ...,"age": ...,"fruits": [...]} and then print a summary: “Name: …,Age: …, Fruits: …”.
name = input("Enter your name: ")
age = int(input("Enter your age: "))
fruits = input("Enter 3 favorite fruits (comma-separated): ")

fruit_list = fruits.split(",")
fruit_list = [f.strip() for f in fruit_list]

user_data = {
    "name": name,
    "age": age,
    "fruits": fruit_list
}

print("\nSummary:")
print("Name:", user_data["name"])
print("Age:", user_data["age"])
print("Fruits:", user_data["fruits"])
```
